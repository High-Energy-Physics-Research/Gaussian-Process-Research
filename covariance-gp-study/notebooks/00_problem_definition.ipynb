# Problem Definition

We study covariance matrix inference from independent zero-mean realizations

$$y^{(m)} \sim \mathcal{N}(0, \Sigma_{true}), \qquad m=1,\ldots,M.$$

The object of interest is $\Sigma_{true}$, not regression of a mean function. The true mean is set to zero in the first experiments to avoid mean inference as a confounder.

Definitions:

- $n$: matrix dimension / number of bins.
- $M$: number of independent realizations available in one experiment.
- $R$: number of Monte Carlo pseudo-experiments used to estimate the distribution of an estimator's error.

The unstructured baseline is

$$\hat{\Sigma}_{sample}=\mathrm{cov}(Y, \mathrm{rowvar}=False, \mathrm{ddof}=1).$$

Structured estimators use a covariance family $\Sigma(\theta)$ and maximize the joint likelihood

$$p(y^{(1)},\ldots,y^{(M)}\mid\theta)=\prod_{m=1}^{M}\mathcal{N}(y^{(m)}\mid 0,\Sigma(\theta)).$$

The main metric is the relative Frobenius error

$$E_F = \frac{\|\hat{\Sigma}-\Sigma_{true}\|_F}{\|\Sigma_{true}\|_F}.$$
